```python
import networkx as nx
import matplotlib.pyplot as plt
import random
import numpy as np
from scipy.stats import pearsonr
from power_law_fit import read_edgelist_from_csv,plot_degree_distribution,fit_powerlaw
from property_evaluation import graph_analysis
from collections import Counter
import Graph_Sampling
import pandas as pd
import os
```

```python
folder_path = 'data/subgraph_data'

# Create an empty list to store the file paths
file_paths = []

# Iterate over all files in the specified folder
for file_name in os.listdir(folder_path):
    # Check if the file name starts with 'rabo'
    if file_name.startswith('rabo'):
        # Add the full file path to the list
        full_path = os.path.join(folder_path, file_name)
        file_paths.append(full_path)

# Print the list of file paths
print(file_paths)
```

```python
subgraph_size = 50000
file_path = 'data/subgraph_data/rabo1.csv'
df = pd.read_csv(file_path)

G_di = nx.from_pandas_edgelist(df,source = 'source', target = 'target', 
                                  edge_attr=True,create_using=nx.DiGraph)

```

```python
file_path = file_paths[0]
df = pd.read_csv(file_path)

G_di = nx.from_pandas_edgelist(df,source = 'source', target = 'target', 
                                  edge_attr=True,create_using=nx.DiGraph)

G_un = nx.Graph(G_di)
print('-------------------------------\nsubgraph:')
print(graph_analysis(G_un))
degrees = [degree for node, degree in G_un.degree()]
fit_powerlaw(degrees)
print()
```
